# Project Amber 简体中文采集记录

本 Notebook 调用与命令行相同的采集核心。采集前会读取 `robots.txt`；若收到 403、429 或验证页会立即停止，不会绕过访问限制。

```python
import site
from pathlib import Path

# 处理 pip install -e . 后尚未重启的 Jupyter 内核。
for site_package in site.getsitepackages():
    site.addsitedir(site_package)

import genshin_text
from rich.console import Console
from genshin_text.collector import CollectionConfig, DEFAULT_CATEGORIES, crawl

project_root = Path(genshin_text.__file__).resolve().parent.parent
console = Console()
config = CollectionConfig(
    db_path=project_root / 'data/genshin.sqlite3',
    raw_dir=project_root / 'data/raw',
    delay_seconds=2.5,
)
DEFAULT_CATEGORIES
```

## 1. 先校验访问规则

此步骤只请求 `robots.txt`，不下载游戏内容。

```python
crawl(config=config, check_only=True, console=console)
```

## 2. 试抓与全量采集

先将 `RUN_TRIAL` 改为 `True`，每个分类最多下载 3 条。确认数据结构和站点访问状态后，再将 `RUN_FULL_COLLECTION` 改为 `True`。断开或停止后重新运行会跳过已完成条目。

```python
RUN_TRIAL = True
RUN_FULL_COLLECTION = True

if RUN_TRIAL:
    crawl(DEFAULT_CATEGORIES, config=config, limit=3, console=console)

if RUN_FULL_COLLECTION:
    crawl(DEFAULT_CATEGORIES, config=config, console=console)

if not (RUN_TRIAL or RUN_FULL_COLLECTION):
    print('采集开关均为 False，未下载游戏文本。')
```
