# 本地检索与打印导出

此 Notebook 不访问网络，只查询 `data/genshin.sqlite3`。

```python
import site
from pathlib import Path
from IPython.display import HTML, display

# 处理 pip install -e . 后尚未重启的 Jupyter 内核。
for site_package in site.getsitepackages():
    site.addsitedir(site_package)

import genshin_text
from genshin_text.exporter import html_document
from genshin_text.store import Store

project_root = Path(genshin_text.__file__).resolve().parent.parent
db_path = project_root / 'data/genshin.sqlite3'
store = Store(db_path)
[(row['category'], row['count']) for row in store.categories()]
```

## 检索

中文检索使用字符子串匹配，并由 SQLite FTS5 预筛选。

```python
query = '璃月'
category = ''  # 例如：'任务/世界任务'
rows = store.search(query, category, limit=50)
[(row['category'], row['title'], row['source_key']) for row in rows]
```

## 生成可打印 HTML

在浏览器打开生成文件后使用打印功能。

```python
output = project_root / 'exports/search-results.html'
output.parent.mkdir(exist_ok=True)
output.write_text(html_document(rows), encoding='utf-8')
display(HTML(f'<a href="../exports/{output.name}" target="_blank">打开可打印 HTML</a>'))
store.close()
```
